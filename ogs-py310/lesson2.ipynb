```python
def my_function(x, y):
    """
    这个自定义函数计算两个数值x和y的和
    函数输入为:
        x: The first number to be added.
        y: The second number to be added.
    函数输出为:
        The sum of x and y.
    """
    return x + y

# 打印结果
print(my_function(1.5, 2))
```
