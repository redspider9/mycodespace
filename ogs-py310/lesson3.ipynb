```python
import matplotlib.pyplot as plt
import numpy as np
from sympy import symbols, lambdify, exp, diff
```

```python
plt.rcParams
```

```python
print(plt.rcParams.get('figure.figsize'))
```

```python
p = plt.rcParams
p["font.sans-serif"] = ["Roboto"]
p["font.weight"] = "light"
p["ytick.minor.visible"] = True
p["xtick.minor.visible"] = True
p["axes.grid"] = True
p["grid.color"] = "0.5"
p["grid.linewidth"] = 0.5
```

```python
x, y = symbols("x, y")
f = exp(-x**2)
f_fcn = lambdify(x, f)

x_array = np.linspace(-3,3,100)
f_array = f_fcn(x_array)

df_dx = diff(f, x)
df_dx_fcn = lambdify(x, df_dx)
df_dx_array = df_dx_fcn(x_array)
```

```python
fig = plt.figure()
ax = fig.add_subplot(1, 1, 1)
ax.plot(x_array, f_array)
ax.set_xlim(-3, 3)
ax.set_ylim(0, 1.2)
# fig.savefig('Figures/默认大小.svg', format='svg')
plt.show()
```

```python
fig = plt.figure(figsize = (3, 2))
# 图片宽度：3 inches；图片高度：2 inches
ax = fig.add_subplot(1, 1, 1)
# 或者 111
ax.plot(x_array, f_array)
ax.set_xlim(-3, 3)
ax.set_ylim(0, 1.2)
plt.xticks(fontname = "Times New Roman")
plt.xticks(fontsize = "8")
plt.yticks(fontname = "Times New Roman")
plt.yticks(fontsize = "8")
plt.show()
# fig.savefig('3 inch x 2 inch.svg', format='svg')
```

```python
cm = 1/2.54  
# 将厘米 cm 转化为 inch

fig = plt.figure(figsize = (9 * cm, 6 * cm))
# 图片宽度：9 厘米；图片高度：6 厘米
ax = fig.add_subplot(1, 1, 1)
# 或者 111
ax.plot(x_array, f_array)
ax.set_xlim(-3, 3)
ax.set_ylim(0, 1.2)
```
