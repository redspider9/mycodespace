```python
x1_array = np.linspace(-3, 3, 200)
x2_array = np.linspace(-3, 3, 200)
xx1, xx2 = np.meshgrid(x1_array, x2_array)
```

```python
ff = xx1 * np.exp(-xx1**2 - xx2**2)
```

```python
ff
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

# 绘制二元函数网格曲面
ax.plot_wireframe(xx1, xx2, ff, rstride=10, cstride=10)

plt.show()
```

```python
t1_array = np.linspace(-6,6,300)
t2_array = np.linspace(-10,10,300)
xt1,xt2 = np.meshgrid(t1_array, t2_array)
```
