## 矩阵分解

###   QR 分解

```python
Q,R = qr(A)
'''
1.第一种分解方式
Q 为正交矩阵
R 为上三角矩阵
A = Q @ R 
多用于求解线性方程组

'''

print(f"Q 正交矩阵：{Q}")
print(f"R 上三角矩阵：{Q}")
print(f"Q @ R 还原A：{Q @ R}")



```

### cholesky 分解

```python
try:
    L = cholesky(A)
    '''
    cholesy 分解，L 为下三角形矩阵
    先决条件：A 为正定对称矩阵,多适合于有限元刚度矩阵的分解
    也叫做平方根法
     
    '''
    print(f"下三角矩阵为：{L}")
    print(f"还原矩阵{L@L.T}")

except Exception as e:
    print("矩阵非正定，无法cholesky 分解")
```

 ## 矩阵微积分运算


```python
import sympy as sp
import numpy as np

'''
标量对向量求导
迹求导
矩阵对标量求导
矩阵指数对t 求导
自动微分 JAX 
'''
```

### 标量对向量求导

```python
x1, x2 = sp.symbols("x1, x2", real = True)
x = sp.Matrix([x1, x2])
A = sp.Matrix([[1, 2], [3,4]])
b = sp .Matrix([1,0])

fx = (A *x -b).T * (A * x - b )
fx = fx[0 , 0]
'''
1.调用sympy 函数求解
2.直接进行矩阵运算（2A.T(Ax -b)）,仅可计算数值解
'''

grad = sp.Matrix([sp.diff(fx, x1), sp.diff(fx, x2)])

print("标量f 对向量x 梯度解析解：")
sp.pprint(grad)


sub_dict = {x1:0.1, x2:0.2}
print("x = [0.1, 0.2] 时梯度数值：")
sp.pprint(grad.subs(sub_dict))

x_val = np.array([0.1, 0.2])
A_np = np.array([[1, 2],[3, 4]])
b_np = np.array([1,0])

g_analysis = 2 * A_np.T @ (A_np @ x_val - b_np)
print(f"x = [0.1, 0.2] 时梯度数值：{g_analysis}")
```

### 迹求导

```python
x11, x12, x21, x22 = sp.symbols("x1, x2, x21, x22", real = True)
X = sp.Matrix([[x11,x12], [x21, x22]])
A_sym = sp.Matrix([[2, 1], [0,3]])

f_tr = (A_sym *X).trace()

dX = sp.Matrix([
    [sp.diff(f_tr, x11), sp.diff(f_tr, x12)],
    [sp.diff(f_tr, x21), sp.diff(f_tr, x22)]
])

print("fx 对迹求导：")
sp.pprint(dX)
```

### 矩阵指数对t 求导

```python
from scipy.linalg import expm
A_mat = np.array([[0, 1],[-1, 0]])
t0 = 0.2
'''
fx = exp Ax
dx = A @ exp Ax
'''
expAt = expm(A_mat * 0.2)
deriv_theory = A_mat @ expAt

print("按公式求导：")
print(deriv_theory)


dt = 1e-6  #有限差分进行计算

expAt_dt = expm(A_mat * (t0 + dt))
deriv_df = (expAt_dt - expAt)/dt

print("使用有限差分法进行求解：")
print(deriv_df)
```

### 自动微分 JAX 

```python
# 适合于大型矩阵求解积分运算，仅输出数值解
import jax
import jax.numpy as jnp

def f_jax(x_vec):
    A_jax = jnp.array([[1,2],[3,4]])
    b_jax = jnp.array([1, 0])
    res = A_jax @ x_vec - b_jax

    return jnp.sum(res**2)
```

```python
grad = jax.grad(f_jax)
xv = jnp.array([0.1, 0.2])
g_jax = grad(xv)

print(f"JAX 自动微积分再（0.1，0.2）处数值解：{g_jax}")
```

## 矩阵叉积点积混合积运算

```python
import numpy as np
import math
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
'''
点积运算有三种
np.dot
@
np.inner
'''
dot1 = np.dot(a, b)
dot2 = a @ b 
dot3 = np.inner(a, b)


norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)
cos_theta = dot1/(norm_a * norm_b)

theta_rad = math.acos(cos_theta)
theta_deg = math.degrees(theta_rad)
print(theta_deg)
print(dot1,dot2,dot3)

'''
叉积仅调用np.cross
叉积交换次序，鞋子正反脚表述方向性的不同
'''

c = np.cross(a,b)
d = np.cross(b,a)
print(c,d)

mix = np.dot(a, np.cross(b, c))
print(mix)
```

## 矩阵特征值特征向量求解

```python
import numpy as np

A = np.array([
    [2,1],
    [1,2]
])

'''
eigvals:特征值lamda
eigvecs: 特征向量x
'''
eigvals,eigvecs = np.linalg.eig(A)
print("特征值：")
print(eigvals)
print("每列为特征向量：")
print(eigvecs)

'''
对特征向量矩阵取值运算
注意np.linalg.eig 输出特征向量矩阵为按列对应
'''
lam0 = eigvals[0]

x0 = eigvecs[:, 0]

left = A @ x0
right = lam0 * x0

print(left)
print(right)
```

### 实对称阵

```python
S = np.array([
    [3,2,1],
    [2,2,1],
    [1,1,1]
])

'''
对于实对称阵，不同特征值的特征向量点积为0
由于特征值是一种矩阵特性
故而实对称阵可以转换为对角矩阵，且是正交阵，可以实现坐标系旋转
1.实对称阵是积于垂直坐标下的读数，变量可以转换为相互独立
'''
vals,vecs = np.linalg.eig(S)
v1 = vecs[:,0]
v2 = vecs[:,1]
print(v1 @ v2)
```

### 病态矩阵判定

```python
 import numpy as np
from scipy.sparse import lil_matrix, csr_matrix
import scipy.sparse.linalg as spla

# ======================
# 位移法示例：3节点梁，每个节点2自由度(竖向位移+转角)，共6自由度
# ======================
ndof = 6
# 用LIL格式方便组装（适合有限元组装）
K_sp = lil_matrix((ndof, ndof))
EI = 1.0
L = 1.0

# 梁单元刚度矩阵（单个单元，2节点，4自由度）
def beam_elem_k(EI, L):
    return np.array([
        [12*EI/L**3,  6*EI/L**2, -12*EI/L**3,  6*EI/L**2],
        [6*EI/L**2,   4*EI/L,    -6*EI/L**2,   2*EI/L],
        [-12*EI/L**3, -6*EI/L**2, 12*EI/L**3, -6*EI/L**2],
        [6*EI/L**2,   2*EI/L,    -6*EI/L**2,   4*EI/L]
    ])

# 单元1：节点0 ↔ 节点1，自由度 [0,1] 和 [2,3]
ke1 = beam_elem_k(EI, L)
dof1 = [0,1,2,3]
for i in range(4):
    for j in range(4):
        K_sp[dof1[i], dof1[j]] += ke1[i,j]

# 单元2：节点1 ↔ 节点2，自由度 [2,3] 和 [4,5]
ke2 = beam_elem_k(EI, L)
dof2 = [2,3,4,5]
for i in range(4):
    for j in range(4):
        K_sp[dof2[i], dof2[j]] += ke2[i,j]

K_csr = K_sp.tocsr()
print("===== 位移法 全局刚度矩阵（稀疏）====")
print("稠密形式查看：")
print(K_csr.toarray())
print(f"非零元素数量：{K_csr.nnz}")
print(f"矩阵总元素：{ndof*ndof}")
print(f"稀疏度(非零占比)：{K_csr.nnz/(ndof*ndof):.3f}\n")

# ======================
# 力法示例：同一根梁，2次超静定，柔度矩阵【稠密】
# ======================
# 假设2个多余未知力 X1,X2，柔度矩阵所有元素都不为0
delta = np.array([
    [0.1667, 0.0833],
    [0.0833, 0.1667]
])
print("===== 力法 柔度矩阵（稠密）====")
print(delta)
print(f"力法矩阵大小：{delta.shape}, 全部元素非零，稠密\n")

# 力法方程 delta @ X = Delta0
Delta0 = np.array([0.02, 0.01])
X = np.linalg.solve(delta, Delta0)
print("力法求解多余未知力 X = ", X)

# 位移法施加边界条件，求解例子
# 约束节点0竖向位移dof0=0，节点2竖向位移dof4=0
K_dense = K_csr.toarray()
F = np.zeros(ndof)
F[2] = -1.0 # 在节点1施加竖向荷载
fix_dofs = [0,4]
for d in fix_dofs:
    K_dense[d,:]=0
    K_dense[:,d]=0
    K_dense[d,d]=1
    F[d]=0

u = np.linalg.solve(K_dense, F)
print("\n位移法求解节点位移 u = ", u)
```
